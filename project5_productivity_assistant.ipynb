# Project 5 — AI Productivity Assistant (Jupyter version)

Same 8 productivity tools as the Streamlit app, exposed here as plain
Python functions so you can call them directly in a notebook.

Uses Google's **current** Gen AI SDK (`google-genai`). The old
`google-generativeai` package is deprecated — support ended Nov 30, 2025.

1. Summarize meeting notes
2. Generate action items
3. Rewrite emails
4. Create presentation outlines
5. Generate LinkedIn posts
6. Brainstorm ideas
7. Translate text
8. Create study notes


## Setup

```
%pip install -q google-genai
```

```
import os
import getpass
from google import genai

api_key = os.environ.get("GEMINI_API_KEY") or getpass.getpass("Enter your Gemini API key: ")
client = genai.Client(api_key=api_key)

MODEL_NAME = "gemini-3.6-flash"

def generate(prompt: str) -> str:
    """Send a prompt to Gemini and return the text response."""
    response = client.models.generate_content(model=MODEL_NAME, contents=prompt)
    return response.text

print("Gemini client ready.")
```

## 1. Summarize Meeting Notes

```
def summarize_meeting_notes(notes: str) -> str:
    prompt = f"Summarize the following meeting notes into clear, concise bullet points:\n\n{notes}"
    return generate(prompt)

# Example:
# print(summarize_meeting_notes("""
# We discussed the Q3 roadmap. Sarah will lead the redesign.
# Launch is targeted for October. Budget approval is pending from finance.
# """))
```

## 2. Generate Action Items

```
def generate_action_items(notes: str) -> str:
    prompt = (
        "Extract clear action items from the text below. For each item, include: "
        "Task, Owner (if mentioned, else 'Unassigned'), and Deadline (if mentioned, "
        "else 'Not specified'). Format as a table.\n\n" + notes
    )
    return generate(prompt)

# Example:
# print(generate_action_items("Sarah will finish the redesign by Friday. John needs to email the vendor."))
```

## 3. Rewrite an Email

```
def rewrite_email(email_text: str, tone: str = "Professional") -> str:
    prompt = f"Rewrite the following email in a {tone.lower()} tone. Keep the core message intact:\n\n{email_text}"
    return generate(prompt)

# Example:
# print(rewrite_email("hey can u send the report today im waiting", tone="Professional"))
```

## 4. Create a Presentation Outline

```
def presentation_outline(topic: str, slides: int = 8) -> str:
    prompt = (
        f"Create a {slides}-slide presentation outline on '{topic}'. "
        "For each slide, give a title and 2-4 bullet points of content."
    )
    return generate(prompt)

# Example:
# print(presentation_outline("Artificial Intelligence in Education", slides=6))
```

## 5. Generate a LinkedIn Post

```
def linkedin_post(topic: str, tone: str = "Professional") -> str:
    prompt = (
        f"You are a professional LinkedIn content writer. Write an engaging "
        f"LinkedIn post in a {tone.lower()} tone about: {topic}. "
        "Include relevant hashtags at the end."
    )
    return generate(prompt)

# Example:
# print(linkedin_post("Finishing my AI capstone project", tone="Inspirational"))
```

## 6. Brainstorm Ideas

```
def brainstorm(topic: str, num_ideas: int = 5) -> str:
    prompt = f"Brainstorm {num_ideas} creative, distinct ideas for: {topic}. Number them."
    return generate(prompt)

# Example:
# print(brainstorm("Capstone project ideas using generative AI", num_ideas=5))
```

## 7. Translate Text

```
def translate(text: str, target_language: str) -> str:
    prompt = f"Translate the following text into {target_language}. Only return the translation:\n\n{text}"
    return generate(prompt)

# Example:
# print(translate("Good morning, how are you?", "Hindi"))
```

## 8. Create Study Notes

```
def study_notes(content_or_topic: str) -> str:
    prompt = (
        "Turn the following content/topic into clear, well-organized study notes "
        "with headings, bullet points, and a short summary at the end:\n\n" + content_or_topic
    )
    return generate(prompt)

# Example:
# print(study_notes("Explain supervised vs unsupervised machine learning"))
```
